```python
 #Fix Date Columns
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)
```

```python
#Invalid Shipping Dates
invalid_dates = df[
    df["ship_date"] < df["order_date"]
]

len(invalid_dates)
```

```python
#clean text column
text_columns = df.select_dtypes(
    include="object"
).columns

for col in text_columns:
    df[col] = (
        df[col]
        .str.replace("\xa0", " ", regex=False)
        .str.strip()
    )
```

```python
#sales Outlier
Q1 = df["sales"].quantile(0.25)
Q3 = df["sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["sales"] < lower_bound) |
    (df["sales"] > upper_bound)
]

len(outliers)
```

# FEATURE ENGINEERING

```python
#Create Shipping Days
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
df["shipping_days"].describe()
```

```python
df
```

```python
# Create Date Features
# ============================================================
# 19. Create Date Features - Feature Engineering
# ============================================================

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
df["quarter"] = df["order_date"].dt.quarter
df["day"] = df["order_date"].dt.day
df["day_of_week"] = df["order_date"].dt.day_name()
df["year_month"] = df["order_date"].dt.to_period("M").astype(str)

# Display the newly created date features
print("=== Date Features Created Successfully ===\n")

print(df[[
    "order_date",
    "year",
    "month",
    "month_name",
    "quarter",
    "day",
    "day_of_week",
    "year_month"
]].head(10))
```

```python
df
```

```python
#Create session
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

df["season"] = df["month"].apply(get_season)

 
```

# KPI & EDA ANALYSIS

```python
# ============================================================
# KPI CALCULATIONS
# ============================================================

total_sales = df["sales"].sum()
total_orders = df["order_id"].nunique()
total_customers = df["customer_id"].nunique()
total_products = df["product_id"].nunique()

average_order_value = total_sales / total_orders

average_sales_per_line = df["sales"].mean()

average_shipping_days = df["shipping_days"].mean()

kpi = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Orders",
        "Total Customers",
        "Unique Products",
        "Average Order Value",
        "Average Sales per Line",
        "Average Shipping Days"
    ],
    
    "Value": [
        total_sales,
        total_orders,
        total_customers,
        total_products,
        average_order_value,
        average_sales_per_line,
        average_shipping_days
    ]
})

kpi
```

```python
 
# ==============================
#  SALES DISTRIBUTION
# ==============================

plt.figure(figsize=(8, 5))

sns.histplot(df["sales"], bins=30, kde=True)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Number of Orders")

plt.show()
```

```python
 
# ==============================
# SALES BY REGION
# ==============================

region_sales = (
    df.groupby("region")["sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Region:")
display(region_sales)

plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
# ==============================
# 7. SALES BY CATEGORY
# ==============================

category_sales = (
    df.groupby("category")["sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Category:")
display(category_sales)

plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
# ==============================
# 9. TOP 10 PRODUCTS
# ==============================

top_products = (
    df.groupby("product_name")["sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print("Top 10 Products:")
display(top_products)

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.show()
```

```python
# ==============================
# 10. BOTTOM 10 PRODUCTS
# ==============================

bottom_products = (
    df.groupby("product_name")["sales"]
      .sum()
      .sort_values()
      .head(10)
)

print("Bottom 10 Products:")
display(bottom_products)

plt.figure(figsize=(10, 6))

bottom_products.plot(kind="barh")

plt.title("Bottom 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.show()
```

```python
# ==============================
# 11. YEARLY SALES TREND
# ==============================

yearly_sales = (
    df.groupby("year")["sales"]
      .sum()
)

print("Yearly Sales:")
display(yearly_sales)

plt.figure(figsize=(8, 5))

yearly_sales.plot(marker="o")

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.grid(True)
plt.show()
```

```python
# ==============================
# 12. MONTHLY SALES TREND
# ==============================

monthly_sales = (
    df.groupby("year_month")["sales"]
      .sum()
)

print("Monthly Sales:")
display(monthly_sales.head())

plt.figure(figsize=(12, 5))

monthly_sales.plot(marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
# ==============================
# 13. MONTH-WISE SALES
# ==============================

month_sales = (
    df.groupby(["month", "month_name"])["sales"]
      .sum()
      .reset_index()
      .sort_values("month")
)

print("Month-wise Sales:")
display(month_sales)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=month_sales,
    x="month_name",
    y="sales"
)

plt.title("Sales by Month")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.show()
```

```python
# ==============================
# 14. QUARTER-WISE SALES
# ==============================

quarter_sales = (
    df.groupby("quarter")["sales"]
      .sum()
)

print("Quarter-wise Sales:")
display(quarter_sales)

plt.figure(figsize=(7, 5))

quarter_sales.plot(kind="bar")

plt.title("Sales by Quarter")
plt.xlabel("Quarter")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
# ==============================
# 15. SEASONAL SALES ANALYSIS
# ==============================

season_sales = (
    df.groupby("season")["sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Season:")
display(season_sales)

plt.figure(figsize=(8, 5))

season_sales.plot(kind="bar")

plt.title("Sales by Season")
plt.xlabel("Season")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
# ==============================
# 16. SALES BY SHIP MODE
# ==============================

ship_sales = (
    df.groupby("ship_mode")["sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Ship Mode:")
display(ship_sales)

plt.figure(figsize=(8, 5))

ship_sales.plot(kind="bar")

plt.title("Sales by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Sales")

plt.xticks(rotation=30)
plt.show()
```

```python
# ==============================
# 17. SHIPPING PERFORMANCE
# ==============================

shipping_summary = (
    df.groupby("ship_mode")["shipping_days"]
      .mean()
      .sort_values()
)

print("Average Shipping Days:")
display(shipping_summary)

plt.figure(figsize=(8, 5))

shipping_summary.plot(kind="bar")

plt.title("Average Shipping Days by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Average Shipping Days")

plt.xticks(rotation=30)
plt.show()
```

```python
# ==============================
# 18. SALES BY CUSTOMER SEGMENT
# ==============================

segment_sales = (
    df.groupby("segment")["sales"]
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Customer Segment:")
display(segment_sales)

plt.figure(figsize=(8, 5))

segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
# ==============================
# 19. REGION × CATEGORY
# ==============================

region_category = pd.pivot_table(
    df,
    values="sales",
    index="region",
    columns="category",
    aggfunc="sum"
)

print("Sales by Region and Category:")
display(region_category)

plt.figure(figsize=(9, 6))

sns.heatmap(
    region_category,
    annot=True,
    fmt=".0f"
)

plt.title("Sales by Region and Category")

plt.show()
```

```python
# ==============================
# 21. TOP 10 CUSTOMERS
# ==============================

top_customers = (
    df.groupby(["customer_id", "customer_name"])["sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print("Top 10 Customers:")
display(top_customers)
```

```python
# ==============================
# 22. TOP 10 PRODUCT CONTRIBUTION
# ==============================

top_10_sales = top_products.sum()

total_product_sales = df["sales"].sum()

contribution = (top_10_sales / total_product_sales) * 100

print(f"Top 10 products contribute {contribution:.2f}% of total sales.")
```

```python
# ==============================
# 23. EDA SUMMARY
# ==============================

print("========== EDA SUMMARY ==========")

print(f"Total Sales: ${df['sales'].sum():,.2f}")
print(f"Total Orders: {df['order_id'].nunique():,}")
print(f"Total Customers: {df['customer_id'].nunique():,}")

print("\nBest Performing Region:")
print(region_sales.idxmax())

print("\nBest Performing Category:")
print(category_sales.idxmax())

print("\nBest Performing Season:")
print(season_sales.idxmax())

print("\nBest Performing Customer Segment:")
print(segment_sales.idxmax())

print("\nTop Product:")
print(top_products.index[0])

print("\nEDA completed successfully!")
```
